The second tutorial covers the Monte Carlo method and is very similar in style to DL_MONTE. Again the program requires the three input files employed in tutorial 1. Basin.xyz should be built as before or you can use the same one as tutorial 1. Also the potentials files can be reused. The control file is slightly different.

The code below will create the control file

method monte
steps 50
eqilsteps 25
temperature 1300
move atoms 1 50
K
move swap 1 50
Al Si
close

In this example, "monte" is used as the keyword for method and "move atoms" is employed to displace the atoms chosen at random. Here we have only moved the K ions to save time whilst simultaneously attempting to swap the Si and Al. It is bit of a trivial example as only the K ions are displaced and The Si/Al swapped, but illustrates how to set up a basic MC simulation.

```python

# Sample code to generate output
sample_list = ['method monte', 'steps 50', 'eqilsteps 25', 'temperature 1300', 'move atoms 1 50', 'K', 'move swap 1 50', 'Al Si', 'close']

for item in sample_list:
    print(item)
# Save the captured output to a text file
file = open('control', 'w')
for item in sample_list:
    file.write(item)
    file.write("\n")

file.flush()
file.close()
```

you will see that the maximum displacement of an atom is 0.001 A, which is too small as all the moves are accepted (not ideal for sampling). As an exercise add the line "maxdistance 0.100" and re-run the simulation. The ratio of accepted:rejected atom translations is giverned by the magnitude of the displacement. The latter is updated as the simulation proceeds; by delut this is every 1000 steps and has a default target ratio of 0.37. Both of these can be adjusted using the keywords distanceupdate and distanceratio respectively.

The above example uses the NVT ensemble. It is also possible to use the NPT ensemble. The following example sets up a much longer simulation using volume moves.

Create the following control file - the potentials and xyz file can remain the same. Warning it will take a reasonably long time

method monte
steps 25000
eqilsteps 1000
distanceupdate 1000
distanceratio 50
temperature 1300
symmetry vectors   #adjust lattice vectors independently during volume moves
move volume 10     #indicates that volume moves are required at
move atoms 1 50
K
move swap 1 50
Al Si
close
