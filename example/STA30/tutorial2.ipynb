# PyMC-NMR Tutorial 2: STA30 Monte Carlo

This tutorial sets up a standard Monte Carlo simulation using the modern YAML configuration. It displaces K ions and swaps Al/Si.

```python
import yaml
from pymc_nmr.monte import run_simulation

mc_config = {
    'job': {
        'structure_path': 'basin.xyz',
        'structure_method': 'monte',
        'temperature': 1300.0,
        'equil_steps': 25
    },
    'potential': {
        'arch': 'mace_mp',
        'model': 'MACE-matpes-r2scan-omat-ft.model',
        'device': 'cpu',
        'precision': 'float64'
    },
    'monte': {
        'steps': 50,
        'atom_move_freq': 50,
        'move_types': ['K']
    },
    'moves': {
        'swaps': [
            {'pairs': [['Al', 'Si']], 'frequency': 50}
        ]
    }
}

with open('config_mc.yaml', 'w') as f:
    yaml.safe_dump(mc_config, f, default_flow_style=False)

run_simulation(config_path='config_mc.yaml', log_path='mc_monte.log')
```

The summary in `mc_monte.log` will report accepted atom moves and accepted Al/Si swaps. If all atom moves are accepted, increase `max_distance` under the `monte` section.
